```python
x=df[['test_score(out of 10)','interview_score(out of 10)']]
y=df['salary($)']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y)
```

```python
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
multi_lin= LinearRegression()
multi_lin.fit(x_train,y_train)
```

```python
predicted_values = multi_lin.predict(x_test)
predicted_values
```

```python
multi_lin.coef_
```

```python
multi_lin.score(x_train,y_train)
```

```python
multi_lin.score(x_test,y_test)
```

```python
pd.DataFrame({"Actual_Data" : y_test,
             "New_Prediction" : predicted_values,
             "Error":y_test - predicted_values
             })
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
MSE = mean_squared_error(y_test,predicted_values)
MAE = mean_absolute_error(y_test,predicted_values)
RMSE = np.sqrt(MSE)
r_squared = r2_score(y_test,predicted_values)
```
